## 9-3 类成员
公有成员：公有成员是指在类的内部和外部都可以访问的变量和方法。在 Python 中，默认情况下，类的所有成员都是公有的

```python
class MyClass:
    def __init__(self):
        self.public_variable = "I am a public variable"
    def public_method(self):
        print("This is a public method")
# 使用类
obj = MyClass()
print(obj.public_variable)  # 访问公有变量
obj.public_method()         # 调用公有方法
```

保护成员：是指仅在类本身及其子类中可访问的变量和方法。

```python
class MyClass:
    def __init__(self):
        self._protected_variable = "I am a protected variable"
    def _protected_method(self):
        print("This is a protected method")

class SubClass(MyClass):
    def demo_method(self):
        print(self._protected_variable)  # 子类中访问保护成员
        self._protected_method()

# 使用子类
obj = SubClass()
obj.demo_method()
```

私有成员：是指只能在类的内部访问的变量和方法。

```python
class MyClass:
    def __init__(self):
        self.__private_variable = "I am a private variable"
    def __private_method(self):
        print("This is a private method")
    def access_private_method(self):
        # 在类内部访问私有方法
        self.__private_method()
# 使用类
obj = MyClass()
# obj.__private_variable  # 这会引发错误，因为这是私有变量
obj.access_private_method()  # 间接访问私有方法
```

## 9-5 方法

```python
class Fish:
    '''鱼类'''

    fish_count = 0  # 一个类变量fish_count

    def __init__(self, id):
        self.id = id   # 一个对象变量id
        Fish.fish_count += 1  # 类变量的调用
        print('creating fish{}'.format(self.id))

    def die(self):
        Fish.fish_count -= 1
        print("fish{} is dead".format(self.id))

    def swim(self):
        print('fish{} is swimming'.format(self.id))

    @classmethod  #用装饰器标明该函数为类方法
    def get_fish_count(cls): #类方法的定义
        print('This school has {} fishes'.format(cls.fish_count))

school = []
for i in range(3):
    fish = Fish(i)
    school.append(fish)
    fish.swim()
Fish.get_fish_count()
school[0].die()       #对象方法调用
Fish.get_fish_count() #类方法调用
```

## 9-6 类的继承
创建一个用户类User，其中包含姓名和年龄属性、以及打印函数和接收消息的函数。

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def printInfo(self):
       print('name:{0}, age:{1}'.format(self.name, self.age))
    def receive(self, msg):
       print('User {0} receive a message:{1}'.format(self.name, msg))
user = User('Jerry', 20)
user.printInfo()  # output：name:Jerry, age:20
```

（1）创建一个空的子类
创建一个管理员用户Admin，继承了User类

（2）在子类中添加构造函数
在子类中可添加构造函数__init__()，但它会覆盖父类的构造函数__init__()。若仍然需要继承父类的构造函数，可在子类的构造函数中显式调用父类构造函数，调用super()函数实现，返回父类/父对象的引用

（3）在子类中添加属性和方法：将子类所特有的属性和方法添加至子类中

```python
class Admin(User):
    def __init__(self, name, age, year):
        super().__init__(name, age)
        self.year = year
    def sendMsg(self, user, msg):
        print('admin {} is sending a message.'.format(self.name))
        user.receive(msg)
    def printInfo(self):     #重写了父类的相应函数
        print('name:{0}, age:{1}, year:{2}'.format(self.name, self.age, self.year))

```

（4）调用子类的方法：创建子类对象，通过子类对象调用子类方法

```python
user = User('Jerry', 20)
admin = Admin('Tom', 22, 2022)
admin.printInfo()
admin.sendMsg(user, 'Let us always meet each other with smile, for the smile is the beginning of love')
user.printInfo()
```

## 9-8 模拟动物园里的动物行为

```python
import random

class Animal:
    def __init__(self, id):
        self.id = id

    def eat(self):
        print(f"NO.{self.id}: {self.__class__.__name__} 正在吃东西。")

    def sleep(self):
        print(f"NO.{self.id}: {self.__class__.__name__} 正在睡觉。")

    def move(self):
        # 抽象方法，留给子类实现
        pass

class Dog(Animal):
    def move(self):
        print(f"NO.{self.id}: {self.__class__.__name__} 正在奔跑。")

class Eagle(Animal):
    def move(self):
        print(f"NO.{self.id}: {self.__class__.__name__} 正在飞行。")

class Fish(Animal):
    def move(self):
        print(f"NO.{self.id}: {self.__class__.__name__} 正在游泳。")

# 随机生成动物的函数
def generate_random_animals(num_animals):
    animals = []
    animal_classes = [Dog, Eagle, Fish]
    for i in range(1, num_animals + 1):
        animal_class = random.choice(animal_classes)  # 随机选择一个动物类
        animals.append(animal_class(i))                  # 使用ID生成动物实例
    return animals
# 随机生成10只动物
zoo = generate_random_animals(10)
# 模拟动物的活动
for animal in zoo:
    animal.eat()
    animal.sleep()
    animal.move()
    print("-" * 20)  # 分隔符用于区分不同动物的活动
```

## 案例1：金融投资
在财经领域，创建一个基本的Investment（投资）类，然后根据不同的投资类型（例如，股票、债券等）扩展出更具体的类。每种投资类型可以有自己的计算收益的方式，体现出多态的特性

```python
class Investment:
    def __init__(self, name, amount):
        self.name = name
        self.amount = amount

    def calculate_return(self):
        pass

class StockInvestment(Investment):
    def __init__(self, name, amount, dividend_rate):
        super().__init__(name, amount)
        self.dividend_rate = dividend_rate

    def calculate_return(self):
        return self.amount * self.dividend_rate

class BondInvestment(Investment):
    def __init__(self, name, amount, interest_rate):
        super().__init__(name, amount)
        self.interest_rate = interest_rate

    def calculate_return(self):
        return self.amount * self.interest_rate

# 创建投资
stock = StockInvestment("Apple", 10000, 0.02)
bond = BondInvestment("US Government", 10000, 0.03)

# 计算回报
print(f"股票 {stock.name} 的回报：{stock.calculate_return()} 元。")
print(f"债券 {bond.name} 的回报：{bond.calculate_return()} 元。")
```

## 案例2：Python 面向对象编程中的多态 —— 支付系统模拟
题目描述：在一个支付系统中，支持多种支付方式（如信用卡、支付宝和微信支付）。每种支付方式都有自己特定的支付逻辑和退款逻辑。设计一个父类来定义支付行为的接口，并通过多态实现不同支付方式的统一调用。

要求：
- 父类需要定义支付和退款的接口。
- 子类需实现父类定义的接口。
- 使用多态设计一个统一的支付处理函数 process_payment，它接受任意支付方式对象并调用其支付和退款方法。

版本一：简单实现版

```python
# 父类
class Payment:
    def pay(self, amount):
        pass

    def refund(self, amount):
        pass

# 子类1：信用卡支付
class CreditCardPayment(Payment):
    def pay(self, amount):
        return f"信用卡支付成功，金额：¥{amount:.2f}"

    def refund(self, amount):
        return f"信用卡退款成功，金额：¥{amount:.2f}"

# 子类2：支付宝支付
class AlipayPayment(Payment):
    def pay(self, amount):
        return f"支付宝支付成功，金额：¥{amount:.2f}"

    def refund(self, amount):
        return f"支付宝退款成功，金额：¥{amount:.2f}"

# 子类3：微信支付
class WeChatPayment(Payment):
    def pay(self, amount):
        return f"微信支付成功，金额：¥{amount:.2f}"

    def refund(self, amount):
        return f"微信退款成功，金额：¥{amount:.2f}"
# 多态调用
def process_payment(payment_method, amount):
    print(payment_method.pay(amount))
    print(payment_method.refund(amount))

# 测试代码
if __name__ == "__main__":
    credit_card = CreditCardPayment()
    alipay = AlipayPayment()
    wechat = WeChatPayment()

    print("测试信用卡支付：")
    process_payment(credit_card, 100.50)

    print("\n测试支付宝支付：")
    process_payment(alipay, 200.75)

    print("\n测试微信支付：")
    process_payment(wechat, 300.00)
```

版本二：使用 抽象方法

```python
from abc import ABC, abstractmethod

# 抽象基类（父类）
class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        """处理支付逻辑"""
        pass

    @abstractmethod
    def refund(self, amount):
        """处理退款逻辑"""
        pass

# 子类1：信用卡支付
class CreditCardPayment(Payment):
    def pay(self, amount):
        return f"信用卡支付成功，金额：¥{amount:.2f}"

    def refund(self, amount):
        return f"信用卡退款成功，金额：¥{amount:.2f}"

# 子类2：支付宝支付
class AlipayPayment(Payment):
    def pay(self, amount):
        return f"支付宝支付成功，金额：¥{amount:.2f}"

    def refund(self, amount):
        return f"支付宝退款成功，金额：¥{amount:.2f}"

# 子类3：微信支付
class WeChatPayment(Payment):
    def pay(self, amount):
        return f"微信支付成功，金额：¥{amount:.2f}"

    def refund(self, amount):
        return f"微信退款成功，金额：¥{amount:.2f}"

# 模拟支付处理器（多态调用）
def process_payment(payment_method, amount):
    print(payment_method.pay(amount))
    print(payment_method.refund(amount))

# 测试多态逻辑
if __name__ == "__main__":
    # 实例化不同的支付方式
    credit_card = CreditCardPayment()
    alipay = AlipayPayment()
    wechat = WeChatPayment()

    print("测试信用卡支付：")
    process_payment(credit_card, 100.50)  # 信用卡支付并退款

    print("\n测试支付宝支付：")
    process_payment(alipay, 200.75)  # 支付宝支付并退款

    print("\n测试微信支付：")
    process_payment(wechat, 300.00)  # 微信支付并退款
```
