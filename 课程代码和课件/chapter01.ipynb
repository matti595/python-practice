## 案例1：猜数字小游戏
编写一个程序，让计算机随机生成一个 1 到 100 之间的整数，学生输入数字进行猜测，程序给出提示，直到猜对为止。

```python
import random

# 生成 1 到 100 之间的随机数
number = random.randint(1, 100)
print("我想了一个 1 到 100 之间的数字，来猜一猜吧！")

while True:
    guess = int(input("请输入你的猜测："))
    if guess < number:
        print("太小了，再试试！")
    elif guess > number:
        print("太大了，再试试！")
    else:
        print("恭喜你，猜对了！")
        break
```

# 案例2：随机整数除法及异常处理代码调试
设计程序生成一组随机整数，并对相邻的两个数字进行除法操作。程序中包含了异常处理，以防止除以零的情况，且通过打印信息展示除法过程中的各个步骤，以便更好地理解和调试代码。实现代码如下：

```python
import random

def divide_numbers(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Division by zero is not allowed.")
        result = None
    return result

def main():
    numbers = [random.randint(0, 10) for _ in range(10)]
    print("Generated numbers:", numbers)

    for i in range(len(numbers) - 1):
        a = numbers[i]
        b = numbers[i + 1]
        print(f"Dividing {a} by {b}...")
        result = divide_numbers(a, b)
        if result is not None:
            print(f"Result: {result}\n")
        else:
            print("Skipping invalid division.\n")

if __name__ == "__main__":
    main()
```
