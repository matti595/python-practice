## 2-2
对于小的整数或字符串（由于 Python 对这些小的、常用的对象进行了内存优化），is 和 == 可能会表现出相同的结果：

```python
a = 10
b = 10
print(a is b)    # 通常会输出True，因为小整数对象被重用
print(a == b)    # 输出True，因为它们的值相等
```

但对于大的整数或更复杂的对象，is 和 == 表现会不同：

```python
c = 1000
d = 1000
print(c is d)    # 通常会输出 False，因为它们是不同的对象
print(c == d)    # 输出 True，因为它们的值相等
```

在 if a is not b 语句中，检查 a 和 b 是否不是同一个对象。例如：

```python
e = [1, 2, 3]
f = [1, 2, 3]
if e is not f:
    print("e and f are not the same object")
```

题目：假设2000年初，个人A在银行存入1万元，年利率为2%；同时，个人B投资于年利率为10%的基金。计算到2030年初，两人各自的本息总额

```python
account_a, account_b = 10000, 10000
interest_rate_a, interest_rate_b = 0.02, 0.1
total_a = account_a * pow((1 + interest_rate_a), 30)
total_b = account_a * pow((1 + interest_rate_b), 30)
print("Total asset of A: ",round(total_a))
print("Total asset of B: ",round(total_b))
```

## 案例1：货币汇率转换
需求描述：某公司经常需要进行外汇交易，特别是将美元转换为人民币。为此，他们需要一个简单的程序来执行这项转换。

```python
# 变量的作用域: USD_TO_RMB是一个全局常数
USD_TO_RMB = 6.5  # 数字: 这是一个常数，表示当前的汇率
# 字符串: 使用字符串提示用户
usd_amount = float(input("请输入要转换的美元数量: "))  # 数字: 数据类型为浮点数
# 内置数值运算符: 进行基本的数学运算
rmb_amount = usd_amount * USD_TO_RMB
# 输出结果
print("{} 美元等于 {:.2f} 人民币".format(usd_amount, rmb_amount))
```

## 案例2：简单的财务分析
需求描述：公司希望能够快速评估其年度财务表现。需要输入去年和今年的销售额和成本，计算出利润率和增长率。

```python
# 获取去年和今年的销售额及成本
sales_last_year = float(input("请输入去年的销售额: "))
sales_this_year = float(input("请输入今年的销售额: "))
cost_last_year = float(input("请输入去年的成本: "))
cost_this_year = float(input("请输入今年的成本: "))

# 计算去年和今年的利润
profit_last_year = sales_last_year - cost_last_year
profit_this_year = sales_this_year - cost_this_year

# 计算利润率和增长率
profit_margin_last_year = (profit_last_year / sales_last_year)
profit_margin_this_year = (profit_this_year / sales_this_year)
growth_rate = ((sales_this_year - sales_last_year) / sales_last_year)

# 输出财务分析结果
print("去年的利润率为: {:.2%}".format(profit_margin_last_year))
print("今年的利润率为: {:.2%}".format(profit_margin_this_year))
print("销售增长率为: {:.2%}".format(growth_rate))
```

## 案例3：放射性同位素衰减
科学家通过对原子核研究发现，所有放射性物质其原子核数目随时间作负指数函数而衰减，称为衰变定律。该定律使用下列公式计算，经过一定时间后放射性元素的剩余量：
$$
N_t = N_0 e^{\frac{-0.693t}{T}}
$$

其中，N_0是放射性物质的初始质量，N_t是经过时间t之后仍然剩余的质量，T是该物质的半衰期。
现有同位素锶-90的初始质量为100mg，半衰期为38.1年，求200年后的剩余质量

```python
from math import exp  # 导入指数函数

# 锶-90的半衰期、初始质量和时间设定
half_life = 38.1
initial_mass = 100
years = 200

# 计算锶-90经过200年后的剩余质量
remaining_mass = initial_mass * exp(-0.693 * years / half_life)
print('Remaining mass of Sr-90:', remaining_mass)
from math import exp  # 导入指数函数

# 锶-90的半衰期、初始质量和时间设定
half_life = 38.1
initial_mass = 100
years = 200

# 计算锶-90经过200年后的剩余质量
remaining_mass = initial_mass * exp(-0.693 * years / half_life)
print('Remaining mass of Sr-90:', remaining_mass)
```
