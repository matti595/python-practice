## 10-4 数组运算
（1）基本数学运算
标量与数组运算

```python
import numpy as np
array = np.array([1, 2, 3, 4])
result_add = array + 10  # 标量加法
result_mul = array * 2    # 标量乘法
print(result_add)
print(result_mul)
```

数组间元素级运算

```python
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
result_sum = array1 + array2  # 数组加法
result_prod = array1 * array2  # 数组乘法
print(result_sum)
print(result_prod)
```

广播机制运算

```python
array = np.array([[1, 2, 3], [4, 5, 6]])
scalar = 10
result_broadcast1 = array + scalar
print(result_broadcast1)

array1 = np.array([[1, 2, 3], [4, 5, 6]])  # 形状(2,3)
array2 = np.array([10, 20, 30])            # 形状(1,3)
result_broadcast2 = array1 + array2
print(result_broadcast2)
```

（3）线性代数运算

```python
import numpy as np
from numpy.linalg import inv, eig

# 矩阵乘法
matrix1 = np.array([[1, 2], [3, 4]])
matrix2 = np.array([[5, 6], [7, 8]])
product1 = np.dot(matrix1, matrix2)  # 方法1
product2 = matrix1 @ matrix2         # 方法2（等价）
print(product1)
```

```python
# 逆矩阵计算
matrix = np.array([[1, 2], [3, 4]])
inverse_matrix = inv(matrix)
print(inverse_matrix)
```

```python
# 特征值与特征向量计算
matrix = np.array([[1, 2], [2, 1]])
eigenvalues, eigenvectors = eig(matrix)
print("特征值:", eigenvalues)
print("特征向量:\n", eigenvectors)
```

## 10-7 Numpy 向量化示例
（1）消除显式循环

```python
import numpy as np
# 消除显式循环
# 循环方式
lst = [1, 2, 3, 4, 5]
result_loop = [x * 2 for x in lst]

# 向量化方式
arr = np.array(lst)
result_vector = arr * 2
print(result_loop)
print(result_vector)
```

（2）逐元素操作批量化

```python
# 逐元素操作批量化
# 标量方式
x = 2
result_scalar = x ** 2

# 向量化方式
arr = np.array([1, 2, 3, 4])
result_vector2 = arr ** 2
print(result_scalar)
print(result_vector2)
```

（3）内置数学函数向量化

```python
# 内置数学函数向量化
arr = np.array([0, np.pi / 2, np.pi])
result_sin = np.sin(arr)
print(result_sin)
```

（4）简化复杂运算逻辑

```python
# 复杂运算简化
# 循环方式
arr_loop = [1, 2, 3, 4]
result_sum_loop = sum([x**2 for x in arr_loop])

# 向量化方式
arr_vec = np.array([1, 2, 3, 4])
result_sum_vec = np.sum(arr_vec ** 2)
print(result_sum_loop)
print(result_sum_vec)
```

## 案例1：线性回归分析

```python
import numpy as np
# （1）导入库并模拟GDP数据
gdp_data = np.array([18.57, 18.71, 18.83, 18.96, 19.08, 19.21, 19.33, 19.45, 19.57, 19.70])

# （2）计算年度增长率
gdp_growth_rate = np.diff(gdp_data) / gdp_data[:-1] * 100
print("GDP 年度增长率:", gdp_growth_rate)

# （3）准备回归模型数据（含偏置项）
X = np.arange(len(gdp_data)).reshape(-1, 1)
y = gdp_data
X_b = np.c_[np.ones((X.shape[0], 1)), X]  # 拼接偏置项

# （4）求解回归系数（文档中省略求解过程，补充标准最小二乘法求解）
theta_best = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
print("回归系数:", theta_best)

# （5）预测未来GDP
X_new = np.array([[len(gdp_data)], [len(gdp_data) + 1]])  # 未来两年
X_new_b = np.c_[np.ones((X_new.shape[0], 1)), X_new]
gdp_predict = X_new_b.dot(theta_best)
print("未来GDP预测:", gdp_predict)
```
