```python
# agg 方法：对DataFrame的列进行聚合操作
# ===================== 1：准备模拟销售数据 =====================
# 构造包含门店、日期、销量、单价的销售数据表（新手易理解的业务场景）
sales_data = {
    "门店": ["A店", "A店", "B店", "B店", "A店", "B店", "C店", "C店"],
    "日期": ["2026-01-01", "2026-01-02", "2026-01-01", "2026-01-02", 
            "2026-01-03", "2026-01-03", "2026-01-01", "2026-01-02"],
    "销量": [50, 65, 30, 45, 70, 55, 20, 35],
    "单价": [10.5, 10.5, 12.0, 12.0, 10.5, 12.0, 15.0, 15.0]
}
df = pd.DataFrame(sales_data)
df
```

```python
# ===================== 1: 多函数聚合 =====================
# 对指定列（销量）同时执行多个聚合函数（求和、均值、最大值、最小值）
# 方式1：列表形式指定多函数
agg_multi_1 = df["销量"].agg(["sum", "mean", "max", "min"])
# 方式2：对多列执行多函数（销量、单价都执行sum和mean）
agg_multi_2 = df[["销量", "单价"]].agg(["sum", "mean"])

# ===================== 2. 按列指定不同聚合函数 =====================
# 核心场景：不同列需要不同的聚合逻辑（比如销量求和，单价求均值和最大值）
# 用字典指定：key=列名，value=聚合函数（单个/多个）
agg_col_spec = df.agg({
    "销量": "sum",          # 销量列：求和
    "单价": ["mean", "max"] # 单价列：求均值+最大值
})

# ===================== 3. 结合分组 + 按列指定聚合 =====================
# 实战高频场景：先按门店分组，再对每组的销量/单价执行不同聚合
group_agg = df.groupby("门店").agg({
    "销量": ["sum", "mean"],  # 每组销量：求和+均值
    "单价": "max"             # 每组单价：求最大值
})

# ===================== 4. 自定义聚合函数 =====================
# 场景：内置函数满足不了需求时，自定义聚合函数（比如求“极差=最大值-最小值”）
# 自定义函数1：求极差
def range_func(x):
    """计算序列的极差（max - min）"""
    return x.max() - x.min()

# 自定义函数2：求中位数绝对偏差（MAD）
def mad_func(x):
    """计算中位数绝对偏差（可理解为“数据离散程度”的指标）"""
    median = x.median()
    return (x - median).abs().median()

# 应用自定义函数
agg_custom = df.agg({
    "销量": [range_func, mad_func],
    "单价": range_func
})
```

```python
print("=== 1. 多函数聚合：对单列执行多个函数 ===")
print(agg_multi_1)
print("\n=== 对多列执行多个函数 ===")
print(agg_multi_2)
print("\n" + "="*50 + "\n")

print("=== 2. 按列指定不同聚合函数 ===")
print(agg_col_spec)
print("\n" + "="*50 + "\n")

print("=== 3. 分组+agg：按门店分组后，按列指定聚合函数 ===")
print(group_agg)
print("\n" + "="*50 + "\n")

print("=== 4. 自定义聚合函数 ===")
print("销量极差 = 销量最大值 - 销量最小值")
print("单价极差 = 单价最大值 - 单价最小值")
print("销量MAD = 销量的中位数绝对偏差（离散程度指标）")
print(agg_custom)
```

```python
import pandas as pd
import string
import random  # 用于生成随机的字母出现次数

# ===================== 步骤1：准备基础数据 =====================
# 1. 生成26个英文字母（a-z）作为完整的字母表
alphabet = list(string.ascii_lowercase)  # ['a','b','c',...,'z']
# 2. 定义文档数量（这里设为5个文档，可自行调整）
doc_num = 100
# 3. 合成“文档ID-字母-出现次数”数据
synthetic_data = []
random.seed(42)

for doc_id in range(1, doc_num + 1):
    # 为每个文档随机选择若干字母（3-8个，模拟文档内容）
    selected_letters = random.sample(alphabet, random.randint(3, 8))
    # 为每个选中的字母随机生成出现次数（1-5次）
    for letter in selected_letters:
        count = random.randint(1, 5)
        synthetic_data.append({
            "doc_id": doc_id,
            "letter": letter,
            "count": count
        })

# 转换成DataFrame（pivot_table的输入格式）
letter_doc_df = pd.DataFrame(synthetic_data)
letter_doc_df
```

```python
# ===================== 步骤2：用pivot_table构造词袋向量 =====================
# 核心：将纵向的“letter”列旋转为横向列，覆盖全部26个字母
bow_vector = letter_doc_df.pivot_table(
    index="doc_id",          # 行：文档ID（每个文档对应一行向量）
    columns="letter",        # 列：26个英文字母（纵向转横向）
    values="count",          # 单元格值：字母在文档中的出现次数
    fill_value=0,            # 未出现的字母填充0（词袋向量核心要求）
    dropna=False             # 强制保留全部26个字母列（即使某字母未出现）
).reindex(columns=alphabet)  # 确保列的顺序是a-z（随机生成后列可能乱序，整理更易读）

# ===================== 步骤3：输出结果 =====================
# print("=== 合成的文档-字母-出现次数数据（前10行）===")
# print(letter_doc_df.head(10))
# print("\n=== 最终的26字母词袋向量（每行是1个文档的向量）===")
display(bow_vector)
```

## 案例：员工信息数据集清洗与整理
假设有一份员工信息数据集，其中包含姓名、年龄、职位、工资等基本信息。由于数据来源复杂，可能存在缺失值、重复数据和格式不一致等问题。

（1）数据准备

```python
import pandas as pd
import numpy as np

# （1）数据准备
data = {
    '姓名': ['张伟', '李娜', '王强', '刘洋', '张伟', None, '陈敏', '赵磊', '杨杰'],
    '年龄': [25, np.nan, 35, 45, 25, 30, np.nan, 55, 60],
    '职位': ['经理', '开发', '开发工程师', '设计', '经理', '经理', '设计师', None, '开发'],
    '工资': [70000, 80000, np.nan, 75000, 70000, 85000, 72000, 90000, np.nan]
}
df = pd.DataFrame(data)
print("原始数据集：")
print(df)
```

（2）数据清洗与整理步骤

```python
# 1）检测缺失数据
missing_data = df.isna().sum()
print("缺失数据统计:\n", missing_data)
```

```python
# 2）处理缺失数据
df.dropna(subset=['姓名'], inplace=True)  # 删除姓名缺失的行
df['年龄'] = df['年龄'].fillna(df['年龄'].mean())  # 填充年龄缺失值为均值
df.dropna(subset=['工资'], inplace=True)  # 删除工资缺失的行
print("处理缺失数据后：")
print(df)
```

```python
# 3）处理重复数据
df.drop_duplicates(inplace=True)
print("处理重复数据后：")
print(df)
```

```python
# 4）规范化职位格式
position_mapping = {
    '经理': '经理',
    '设计': '设计师',
    '设计师': '设计师',
    '开发': '开发工程师',
    '开发工程师': '开发工程师'
}
df['职位'] = df['职位'].map(position_mapping)
print("规范化职位格式后：")
print(df)
```

（3）最终数据检查

```python
print("清洗后的数据集:\n", df)
print("数据类型:\n", df.dtypes)
```
