## 7-1 读取文本文件
方法1：readlines() 读取所有行（可能占用大量内存）

```python
file = open('./a.txt', 'r',encoding='utf-8')
for line in file.readlines():
    print(line)
file.close()
```

方法2：逐行遍历文件（优化内存占用）

```python
file = open('./a.txt', 'r',encoding='utf-8')
for line in file:
    print(line)
file.close()
```

## 7-2 写入文本文件

```python
# 写入字符串列表，并重定位文件指针读取内容
file = open('./b.txt', 'w+')
list1 = ['Beautiful is better than ugly.', 'Explicit is better than implicit.', 'Simple is better than complex.']
file.writelines(list1)
file.seek(0)  # 定位到文件开头
for line in file:
    print(line)
file.close()
```

## 7-3 二进制文件操作

```python
# 字符串与字节序列转换，读写二进制文件
str_data = "Hello, Python!"
bytes_data = str_data.encode('utf-8')  # 字符串转字节序列

# 写入二进制文件
with open('example.bin', 'wb') as file:
    file.write(bytes_data)

# 读取二进制文件并解码
with open('example.bin', 'rb') as file:
    bytes_data_read = file.read()
decoded_str = bytes_data_read.decode('utf-8')
print(decoded_str)
```

## 7-4 序列化与反序列化（pickle 模块）

```python
import pickle

# 序列化：Python对象写入文件
flowerlist = ['rose', 'tulip', 'sunflower', 'morning glory']
filename = 'flowlist.data'
file = open(filename, 'wb')  # 二进制写入模式
pickle.dump(flowerlist, file)
file.close()

# 反序列化：从文件读取并恢复Python对象
file = open(filename, 'rb')  # 二进制读取模式
storedflowerlist = pickle.load(file)
print(storedflowerlist)
file.close()
```

## 7-5 文件操作（os 与 os.path 模块）

```python
import os

# 检查文件是否存在
if os.path.exists('example.txt'):
    print('文件存在')

# 重命名文件（先判断是否为文件）
if os.path.isfile('old_name.txt'):
    os.rename('old_name.txt', 'new_name.txt')

# 列出当前目录下所有文件和目录
for item in os.listdir('.'):
    print(item)
```

## 7-6  目录操作
（1）列出当前目录下所有 Python 文件

```python
import os

current_dir = os.getcwd()  # 获取当前工作目录
python_files = [f for f in os.listdir(current_dir) if os.path.isfile(f) and f.endswith('.py')]
print(python_files)
```

（2）遍历目录树（打印所有文件和目录路径）

```python
import os

for root, dirs, files in os.walk("./"):
    for name in files:
        print(os.path.join(root, name))  # 打印文件完整路径
    for name in dirs:
        print(os.path.join(root, name))  # 打印目录完整路径
```

## 7-7 CSV 文件处理
（1）读取 CSV 文件

```python
import csv

with open('example.csv', 'r', newline='', encoding='utf-8') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)  # 每行数据以列表形式输出
```

（2）写入 CSV 文件

```python
import csv

with open('output.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['Header1', 'Header2'])  # 写入单行（表头）
    writer.writerows([['data1', 'data2'], ['data3', 'data4']])  # 写入多行
```

## 7-8 JSON 文件处理

```python
import json

# 1. 读取JSON文件
with open('users.json', 'r',encoding='utf-8') as file:
    users = json.load(file)  # 反序列化为Python对象（列表/字典）

# 2. 处理数据（增加年龄、添加兴趣爱好）
for user in users:
    user["age"] += 1
    if "coding" not in user["hobbies"]:
        user["hobbies"].append("coding")

# 3. 写入更新后的数据到新JSON文件
with open('updated_users.json', 'w') as file:
    json.dump(users, file, indent=4)  # indent=4优化格式可读性
```
